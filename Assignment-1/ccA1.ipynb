```python
n = int(input("Enter a number for n: "))
sum = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        sum += i

print(f"The sum of numbers divisible by both 7 and 9 from 1 to {n} is: {sum}")
```

```python
def sum(n):
    sum = 0
    for num in range(2, n + 1):
        is_prime = True
        if num > 1:
            for i in range(2, int(num**0.5) + 1):
                if (num % i) == 0:
                    is_prime = False
                    break
        else:
            is_prime = False

        if is_prime:
            sum += num
    return sum

n = int(input("Enter a number for n: "))
result = sum(n)
print(f"The sum of all prime numbers from 1 to {n} is: {result}")
```
