```python
import numpy as np
arr=np.array([10,20,30,40,50])
print(arr)
print(arr+2)
print(arr*3)
print(arr/2)
```

```python
arr=np.array([1,2,3,4,5])
print(arr)
print(arr[::-1])

x=np.array([1,2,3,4,5,1,2,1,1,1])
values,counts=np.unique(x,return_counts=True)
most_frequent=values[np.argmax(counts)]
indices=np.where(x==most_frequent)[0]
print(most_frequent)
print(indices)

y=np.array([1,1,1,2,3,4,2,4,3,3])
values,counts=np.unique(y,return_counts=True)
most_frequent=values[np.argmax(counts)]
indices=np.where(y==most_frequent)[0]
print(most_frequent)
print(indices)
```

```python
Divyanshi=np.linspace(10,100,25)
print(Divyanshi)
print(Divyanshi.ndim)
print(Divyanshi.shape)
print(Divyanshi.size)
print(Divyanshi.dtype)
print(Divyanshi.nbytes)
transpose_array=Divyanshi.reshape(25,1)
print(transpose_array)
print(Divyanshi.T)
```

```python
ucs420_Divyanshi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print("Original Array:")
print(ucs420_Divyanshi)
print("Mean:", np.mean(ucs420_Divyanshi))
print("Median:", np.median(ucs420_Divyanshi))
print("Maximum:", np.max(ucs420_Divyanshi))
print("Minimum:", np.min(ucs420_Divyanshi))
print("Unique elements:", np.unique(ucs420_Divyanshi))

reshaped_ucs420_Divyanshi = ucs420_Divyanshi.reshape(4, 3)

print("Reshaped Array:")
print(reshaped_ucs420_Divyanshi)

resized_ucs420_Divyanshi = np.resize(ucs420_Divyanshi, (2, 3))

print("Resized Array:")
print(resized_ucs420_Divyanshi)
```
