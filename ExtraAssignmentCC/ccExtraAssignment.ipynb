```python
def rating_category(r):
    if r >= 4.5:
        return "Excellent"
    elif r >= 4.0:
        return "Good"
    else:
        return "Average"

df["Rating_Category"] = df["Performance_Rating"].apply(rating_category)
df[["Name", "Performance_Rating", "Rating_Category"]]
```

```python
ids=df["ID"].to_numpy()
print("Original IDs:", ids)
print("Reversed:", ids[::-1])
```

```python
stress=df["Stress_Level"].to_numpy()
counts=np.bincount(stress)
most_frequent_value=counts.argmax()
indices=np.where(stress == most_frequent_value)[0]
print(stress)
print(most_frequent_value)
print(indices)
```
